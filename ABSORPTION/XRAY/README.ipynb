### `absorption_Xray.py`
- Purpose: Identifies and measures absorption features within the designated speed range v > 0.1c in orger to identify EHVOs.
- Inputs: csv file with all the spectra file names and other info to loop through and folder pathway where spectra files are located.
- Outputs: pdf with plots containing shading for identified features, .txt and csv file with calculated information: BI, EW, depth, and vmin/vmax.

### `x-ray_recon_comparison.py`
- Purpose: To plot and compare spectra and reconstructions provided from Hiremath+2025 with spectra downloaded from SDSS. Data from Hiremath+2025 cuts off generally between -60,000 and -70,000 km/s.
- Inputs: csv file with all the spectra file names and other info to loop through and folder pathway where spectra files are located.
- Outputs: pdf files comparing different lines of data: reconstruction (Hiremath+2025), spectra(Hiremath+2025) and SDSS downloaded spectra - reconstruction (Hiremath+2025), spectra(Hiremath+2025) - normalized/morphed reconstruction (Hiremath+2025), spectra(Hiremath+2025)

## Workflow

1. Download spectra in `SDSS_allepoch_spectra.zip` and `Recons_Hiremath2025.tar.gz` from [Google Drive folder](https://drive.google.com/drive/folders/1u0P_v7Se_zImhhf8_lhY7vy_KjITCvQc?usp=drive_link)
2. Apply filtering (SNR, redshift) using `xray_z_SNR_cutoff.py` to get parent sample
3. Perform X-ray absorption analysis with `absorption_Xray.py` using spectra files in `Recons_Hiremath2025/`
4. Perform visual analysis on plot outputs produced to `OUTPUT_FILES/` to classify absorption features
5. Match datasets using `CSV_NAME_MATCH.py` (OPTIONAL: only if using SDSS file type other than allepoch to compare to Hiremath+2025 reconstructions)
6. Compare reconstructions to downloaded SDSS spectra using `x-ray_recon_comparison.py` to confirm and check for hidden EHVOs
